# Start Spark Master (With Full Spark Binaries)
## Make sure you restarted Jupyter after setting SPARK_HOME

```python
import os
import subprocess

# Set SPARK_HOME if not already set
if 'SPARK_HOME' not in os.environ:
    os.environ['SPARK_HOME'] = 'C:/spark'
    os.environ['PATH'] = f"C:/spark/bin;{os.environ['PATH']}"
    print("Set SPARK_HOME for this session")
else:
    print(f"SPARK_HOME: {os.environ['SPARK_HOME']}")
```

```python
# Start Master using spark-class
print("Starting Spark Master...")
print("="*60)
print("Master URL: spark://192.168.1.5:7077")
print("Web UI: http://localhost:8080")
print("="*60)
print("\nShare with teammates: spark://192.168.1.5:7077")
print("\nThis will keep running. Stop the cell when done.\n")

# Start master
master_cmd = r'C:\spark\bin\spark-class.cmd org.apache.spark.deploy.master.Master --host 192.168.1.5 --port 7077 --webui-port 8080'

try:
    subprocess.run(master_cmd, shell=True)
except KeyboardInterrupt:
    print("\nMaster stopped.")
```
