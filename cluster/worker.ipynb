```python
# Configuration - Update if needed
MASTER_URL = "spark://192.168.1.5:7077"
WORKER_MEMORY = "2g"
WORKER_CORES = "2"
```

```python
# Test connection to master
import socket

master_ip = "192.168.1.5"
master_port = 7077

print(f"Testing connection to master at {master_ip}:{master_port}...")

try:
    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    sock.settimeout(5)
    result = sock.connect_ex((master_ip, master_port))
    sock.close()
    
    if result == 0:
        print("✓ Can reach master!")
    else:
        print("✗ Cannot reach master")
        print("Make sure:")
        print("  1. Master is running")
        print("  2. You're on the same network")
        print("  3. Firewall allows connection")
except Exception as e:
    print(f"✗ Error: {e}")
```

```python
# Start Worker Node
from pyspark.sql import SparkSession
import time

print("="*60)
print("Starting Spark Worker Node")
print("="*60)
print(f"Connecting to: {MASTER_URL}")
print(f"Worker Memory: {WORKER_MEMORY}")
print(f"Worker Cores: {WORKER_CORES}")
print("="*60)

try:
    spark = SparkSession.builder \
        .appName("WorkerNode") \
        .master(MASTER_URL) \
        .config("spark.executor.memory", WORKER_MEMORY) \
        .config("spark.executor.cores", WORKER_CORES) \
        .getOrCreate()
    
    print("\n✓ Worker connected to master!")
    print("✓ This worker is now part of the cluster")
    print("\nKeep this cell running!")
    print("Stop the cell when done.\n")
    
    # Keep worker alive
    while True:
        time.sleep(10)
        
except KeyboardInterrupt:
    print("\nStopping worker...")
    spark.stop()
    print("Worker stopped.")
except Exception as e:
    print(f"\n✗ Error: {e}")
    print("\nTroubleshooting:")
    print("  1. Make sure master is running")
    print("  2. Check network connectivity")
    print("  3. Verify master URL is correct")
    print("  4. Check firewall settings")
```
